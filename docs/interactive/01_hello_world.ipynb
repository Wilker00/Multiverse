## 🧠 **What is Reinforcement Learning?**

RL is learning through interaction. An **agent** makes decisions in an **environment** to maximize **rewards**.

Think of it like training a dog:
- **Agent**: The dog
- **Environment**: Your house/training field
- **Actions**: Sit, stay, fetch, etc.
- **Rewards**: Treats for good behavior
- **Goal**: Dog learns to behave well to get treats

## 🎮 **Our First Environment: Line World**

Line World is a simple 5-position line:
```
Position: 0 ── 1 ── 2 ── 3 ── 4
Reward:   -1   0   0   0   +1
```

- Agent starts at position 2
- Can move left (-1) or right (+1)
- Goal: Reach position 4 (+1 reward) or avoid position 0 (-1 reward)

Let's see it in action!

````python
# First, let's verify Multiverse is installed and working
import subprocess
import sys

def run_command(cmd):
    """Helper to run shell commands"""
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if result.returncode != 0:
        print(f"Error: {result.stderr}")
        return None
    return result.stdout

# Check if multiverse is available
print("Checking Multiverse installation...")
result = run_command("multiverse doctor")
if result:
    print("✅ Multiverse is ready!")
    print(result)
else:
    print("❌ Multiverse not found. Please install it first.")
    print("Run: pip install -r requirements.txt")
````

## 🏃 **Your First Training Run**

Let's train a Q-learning agent on Line World. This will take about 30 seconds.

````python
# Train a Q-learning agent for 50 episodes
print("🚀 Training Q-learning agent on Line World...")
print("This will take about 30 seconds...")

result = run_command("multiverse train --algo q --verse line_world --episodes 50 --quiet")

if result:
    print("✅ Training completed!")
    print("\nTraining output:")
    print(result)
else:
    print("❌ Training failed. Check the error above.")
````

## 📊 **Analyzing Results**

Let's examine what the agent learned.

````python
# Get information about the latest run
print("📈 Analyzing the latest training run...")

result = run_command("multiverse runs latest")
if result:
    print("Run information:")
    print(result)
else:
    print("❌ Could not get run information.")
````

````python
# Let's look at the learning curve
print("📊 Generating learning curve...")

result = run_command("multiverse analyze rewards --run latest")
if result:
    print("Learning curve data:")
    print(result)
else:
    print("❌ Could not generate learning curve.")
````

## 🎬 **Watching the Agent**

Let's simulate the trained agent to see how it behaves.

````python
# Simulate the trained agent
print("🎮 Simulating trained agent behavior...")

result = run_command("multiverse sim --verse line_world --algo q --episodes 3")
if result:
    print("Simulation results:")
    print(result)
else:
    print("❌ Simulation failed.")
````

## 🧪 **Experiment Time!**

Try modifying the training parameters and see what happens.

````python
# Experiment 1: Train longer (100 episodes)
print("🧪 Experiment 1: Training for 100 episodes...")

result = run_command("multiverse train --algo q --verse line_world --episodes 100 --quiet")
if result:
    print("✅ Longer training completed!")
    
    # Check the results
    result2 = run_command("multiverse analyze rewards --run latest")
    if result2:
        print("\n📈 Learning curve (100 episodes):")
        print(result2)
else:
    print("❌ Experiment failed.")
````

````python
# Experiment 2: Different algorithm (SARSA)
print("🧪 Experiment 2: Training with SARSA algorithm...")

result = run_command("multiverse train --algo sarsa --verse line_world --episodes 50 --quiet")
if result:
    print("✅ SARSA training completed!")
    
    # Compare with Q-learning
    result2 = run_command("multiverse analyze rewards --run latest")
    if result2:
        print("\n📈 SARSA learning curve:")
        print(result2)
else:
    print("❌ Experiment failed.")
````

## 🧠 **Understanding What Happened**

### **The RL Loop**
1. **Observe**: Agent sees current state (position on line)
2. **Choose Action**: Agent picks left or right
3. **Environment Responds**: Agent moves, gets reward
4. **Learn**: Agent updates knowledge based on experience
5. **Repeat**: Back to step 1

### **Q-Learning Intuition**
- **Q-Table**: Agent learns Q(position, action) = expected future reward
- **Learning Rule**: Q(s,a) ← Q(s,a) + α[r + γ max_a' Q(s',a') - Q(s,a)]
- **Result**: Agent learns optimal policy (always go right from position 2)

### **Key Insights**
- **Learning Curves**: Show improvement over time
- **Exploration**: Agent tries both actions initially
- **Exploitation**: Agent uses learned knowledge later
- **Convergence**: Performance stabilizes as learning completes
