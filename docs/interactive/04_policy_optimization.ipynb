## Two Approaches to RL

**Value-based** (what we did in Tutorial 03):
- Learn how good each (state, action) is
- Derive policy from values: always pick the highest-value action
- Works well for discrete action spaces

**Policy-based** (this tutorial):
- Directly learn the policy: a mapping from states to actions
- Optimize the policy by gradient ascent on expected reward
- Works for continuous actions, stochastic policies, large action spaces

```
Value-based:  state → Q-values → argmax → action
Policy-based: state → policy(state) → action (directly)
```

### Policy Gradient Intuition

1. Roll out the current policy (collect experience)
2. Compute which actions led to good outcomes
3. Adjust the policy to make good actions more likely
4. Repeat

The **gradient** tells us which direction to adjust the policy parameters.

### Why PPO?

Basic policy gradient has a problem: if you update too aggressively, the policy collapses.
PPO (Proximal Policy Optimization) adds a **clipping constraint** that limits how much the policy can change per update — making training stable.

## Setup

````python
import subprocess

def run(cmd):
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    output = result.stdout + result.stderr
    print(output if output.strip() else '(no output)')
    return result.returncode == 0

run('multiverse doctor')
````

## Part 1: Q-Learning as Baseline

First, establish a Q-learning baseline on grid world so we have something to compare against.

````python
print('Baseline: Q-Learning on grid_world')
run('multiverse train --algo q --verse grid_world --episodes 200 --seed 99')
````

## Part 2: PPO on the Same Task

Now train PPO on the same environment and seed.
PPO uses a neural network policy instead of a Q-table.

````python
print('PPO on grid_world')
run('multiverse train --algo ppo --verse grid_world --episodes 200 --seed 99')
````

**Compare the outputs.** You might notice:
- PPO can be slower to start (it needs enough data to estimate gradients)
- PPO often achieves higher final performance on complex tasks
- Q-learning is faster to converge on simple discrete tasks

## Part 3: Where PPO Shines — Complex Environments

For harder environments with larger state spaces, PPO's neural network policy generalizes better than Q-tables.

````python
# Labyrinth: larger, more complex navigation task
print('Q-Learning on labyrinth_world:')
run('multiverse train --algo q --verse labyrinth_world --episodes 300 --seed 42')
````

````python
print('PPO on labyrinth_world:')
run('multiverse train --algo ppo --verse labyrinth_world --episodes 300 --seed 42')
````

## Part 4: DQN — Best of Both Worlds

DQN (Deep Q-Network) combines Q-learning's stability with neural network generalization:
- Still value-based (learns Q-values)
- Uses a neural network to approximate Q(s,a) instead of a table
- Works where Q-tables are too large

````python
print('DQN on grid_world:')
run('multiverse train --algo dqn --verse grid_world --episodes 200 --seed 99')
````

````python
print('DQN on labyrinth_world:')
run('multiverse train --algo dqn --verse labyrinth_world --episodes 300 --seed 42')
````

## Algorithm Comparison Summary

| Algorithm | Type | Best For | Weakness |
|---|---|---|---|
| Q-Learning | Value-based (table) | Small discrete spaces | Doesn't scale |
| SARSA | Value-based (table) | Safer online learning | Doesn't scale |
| DQN | Value-based (neural net) | Larger discrete spaces | Sample inefficient |
| PPO | Policy-based (neural net) | Complex / continuous tasks | Slow to start |

**Rule of thumb**:
- Small, discrete state space → Q-learning or SARSA
- Large/complex state space, discrete actions → DQN
- Continuous actions or complex policies → PPO
