## The Core Dilemma

Imagine you move to a new city. You found one good restaurant on day one.

- **Exploit**: Go to that restaurant every day (safe, but you might miss better options)
- **Explore**: Try different restaurants (risky, but you might find somewhere much better)

RL agents face the same dilemma every single step.

### Epsilon-Greedy: The Simple Solution

```
epsilon = 0.1  →  10% of the time: random action  (explore)
                  90% of the time: best known action (exploit)
```

Start with high epsilon (explore a lot), decay it over time (exploit more as you learn).

## Setup

````python
import subprocess
import sys

def run(cmd, cwd=None):
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True, cwd=cwd)
    output = result.stdout + result.stderr
    print(output if output.strip() else '(no output)')
    return result.returncode == 0

# Confirm Multiverse is available
run('multiverse doctor')
````

## Part 1: Too Much Exploitation (Greedy Agent)

A fully greedy agent (epsilon=0) always picks the best action it currently knows.
Early in training, it knows almost nothing — so it gets stuck.

We'll use **Grid World**: a 2D grid where the agent navigates to a goal.

````python
# Train a near-greedy agent: very low exploration
# epsilon=0.01 means it almost never explores
print('Training near-greedy agent (epsilon=0.01)...')
run('multiverse train --algo q --verse grid_world --episodes 100 --seed 42 -- --epsilon 0.01')
````

## Part 2: Too Much Exploration (Random Agent)

A fully random agent (epsilon=1.0) always picks random actions.
It never uses what it has learned.

````python
# Train a fully random agent: pure exploration, no exploitation
print('Training fully random agent (epsilon=1.0)...')
run('multiverse train --algo random --verse grid_world --episodes 100 --seed 42')
````

## Part 3: Balanced Exploration (Standard Q-Learning)

Standard Q-learning starts with epsilon=0.3 and decays it over time.
The agent explores aggressively at first, then settles into what it learned.

````python
# Train with balanced exploration — the default Q-learning settings
print('Training with balanced exploration (Q-learning defaults)...')
run('multiverse train --algo q --verse grid_world --episodes 100 --seed 42')
````

## Part 4: Longer Training — Watch the Transition

With epsilon decay, early episodes are exploratory and messy.
Later episodes should be smooth and optimal.
Run more episodes to see the full arc.

````python
# 300 episodes to clearly see the explore → exploit transition
print('Training for 300 episodes to see the full exploration arc...')
run('multiverse train --algo q --verse grid_world --episodes 300 --seed 42')
````

## Experiment: Different Verses, Same Problem

Exploration matters more in some environments than others.
Try the same agent on a maze — which has many dead-ends and requires more exploration.

````python
# Maze world: more complex, requires more exploration to find the path
print('Training on maze_world...')
run('multiverse train --algo q --verse maze_world --episodes 200 --seed 42')
````

````python
# Try cliff_world — a classic exploration problem where wrong exploration is dangerous
print('Training on cliff_world...')
run('multiverse train --algo q --verse cliff_world --episodes 200 --seed 42')
````

## What You Learned

| Setting | Behavior | Problem |
|---|---|---|
| epsilon = 0.0 | Pure greedy | Gets stuck in local optima |
| epsilon = 1.0 | Pure random | Never improves |
| epsilon decay | Explore first, then exploit | Learns well over time |

**Key insight**: The right amount of exploration depends on how complex the environment is.
Harder environments (maze, cliff) need more exploration to find the reward signal at all.
